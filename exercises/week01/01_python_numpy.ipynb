## 1.1 函数

```python
def make_target(x, slope=3, intercept=2):
    y = slope * x + intercept
    return y


result = make_target(4)
print(result)
```

```python
print(make_target(4, slope=5))
print(make_target(4, intercept=-1))
```

```python
import numpy as np

x = np.array([
    [1.0],
    [2.0],
    [3.0],
], dtype=np.float32)

y = make_target(x)

print("x：\n", x)
print("y：\n", y)
print("x 的形状：", x.shape)
print("y 的形状：", y.shape)
print("数据类型：", x.dtype)
```
